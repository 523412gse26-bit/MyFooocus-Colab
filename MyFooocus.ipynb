```python
import sys
import subprocess
import os

print("Python:", sys.version)

# Install dependencies
subprocess.run([
    sys.executable, "-m", "pip", "install", "-U",
    "numpy",
    "cupy-cuda12x",
    "pygit2"
], check=True)

# Download Fooocus
FOOCUS_DIR = "/content/Fooocus"

if not os.path.exists(FOOCUS_DIR):
    subprocess.run([
        "git", "clone",
        "https://github.com/lllyasviel/Fooocus.git",
        FOOCUS_DIR
    ], check=True)

os.chdir(FOOCUS_DIR)

# Add required versions
with open("requirements_versions.txt", "a") as f:
    f.write("\nfastapi==0.104.1")
    f.write("\nstarlette==0.27.0")
    f.write("\npydantic==1.10.19")

# Start Fooocus
subprocess.run([
    sys.executable,
    "entry_with_update.py",
    "--share",
    "--always-high-vram",
    "--all-in-fp16",
    "--listen"
])
```
